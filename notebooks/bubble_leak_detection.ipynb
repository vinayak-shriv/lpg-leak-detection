```python
import os, shutil

# Create proper YOLO folder structure
os.makedirs("/content/proper/images/train", exist_ok=True)
os.makedirs("/content/proper/images/val", exist_ok=True)
os.makedirs("/content/proper/labels/train", exist_ok=True)
os.makedirs("/content/proper/labels/val", exist_ok=True)

# Get all labeled images from test/images
all_images = os.listdir("/content/dataset/test/images")
all_labels = os.listdir("/content/dataset/test/labels")

print(f"Total images: {len(all_images)}")
print(f"Total labels: {len(all_labels)}")

# Split 80% train, 20% val
split = int(len(all_images) * 0.8)
train_imgs = all_images[:split]
val_imgs   = all_images[split:]

# Copy images
for f in train_imgs:
    shutil.copy(f"/content/dataset/test/images/{f}", f"/content/proper/images/train/{f}")
for f in val_imgs:
    shutil.copy(f"/content/dataset/test/images/{f}", f"/content/proper/images/val/{f}")

# Copy matching labels
for f in train_imgs:
    lbl = f.replace(".jpg", ".txt").replace(".png", ".txt")
    src = f"/content/dataset/test/labels/{lbl}"
    if os.path.exists(src):
        shutil.copy(src, f"/content/proper/labels/train/{lbl}")

for f in val_imgs:
    lbl = f.replace(".jpg", ".txt").replace(".png", ".txt")
    src = f"/content/dataset/test/labels/{lbl}"
    if os.path.exists(src):
        shutil.copy(src, f"/content/proper/labels/val/{lbl}")

print(f"Train: {len(os.listdir('/content/proper/images/train'))} images")
print(f"Val:   {len(os.listdir('/content/proper/images/val'))} images")
print(f"Train labels: {len(os.listdir('/content/proper/labels/train'))}")
print(f"Val labels:   {len(os.listdir('/content/proper/labels/val'))}")
```
